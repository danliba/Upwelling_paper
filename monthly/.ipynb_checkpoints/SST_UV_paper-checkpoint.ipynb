## ENSO

```python
import datetime

def anomaly(da, dim='time'):
    clim = da.groupby(f'{dim}.month').mean(dim=dim)
    anom = da.groupby(f'{dim}.month') - clim
    
    return anom
    
NINO = [
    (datetime.datetime(1982, 8, 1), datetime.datetime(1983, 12, 31)),
    (datetime.datetime(1997, 4, 1), datetime.datetime(1998, 6, 30)),
]

def keep_en_events(da, en_windows=NINO, dim='time'):
    """Mask out El Niño periods from a time series."""
    mask = np.zeros(da.sizes[dim], dtype=bool)
    time = da[dim].values
    for start, end in en_windows:
        start64 = np.datetime64(start)
        end64   = np.datetime64(end)
        mask |= ((time >= start64) & (time <= end64))
    return da.isel({dim: mask})

def drop_en_events(da, en_windows=NINO, dim='time'):
    """Mask out El Niño periods from a time series."""
    mask = np.ones(da.sizes[dim], dtype=bool)
    time = da[dim].values
    for start, end in en_windows:
        start64 = np.datetime64(start)
        end64   = np.datetime64(end)
        mask &= ~((time >= start64) & (time <= end64))
    return da.isel({dim: mask})
```

# Now the Heat budget of the MLD

```python
from scipy.io import loadmat
def coastal_band_mask(ds, var='u10', target_width_km=111):
    """
    Create a coastal band mask with uniform cross-shore width in km.
    """
    da = ds[var]
    if 'time' in da.dims:
        da = da.mean(dim='time')

    valid2d = np.isfinite(da).values.astype(bool)
    lat_vals = ds['lat'].values
    lon_vals = ds['lon'].values
    dlon = np.abs(lon_vals[1] - lon_vals[0])  # grid spacing in degrees

    # Find coastline (easternmost valid cell per row)
    band = np.zeros_like(valid2d, dtype=bool)

    for j in range(valid2d.shape[0]):
        row = valid2d[j, :]
        ocean_cols = np.where(row)[0]
        if len(ocean_cols) == 0:
            continue

        # Coast = easternmost valid cell
        i_coast = ocean_cols[-1]

        # How many cells = target_width_km at this latitude?
        A = 6371.0  # km
        dx_km = A * np.cos(np.deg2rad(lat_vals[j])) * np.deg2rad(dlon)
        n_cells = int(np.round(target_width_km / dx_km))

        # Fill n_cells westward from coast
        i_start = max(i_coast - n_cells, 0)
        band[j, i_start:i_coast + 1] = True

    # Only keep cells that are actually ocean
    band = band & valid2d

    da_mask = xr.DataArray(band, dims=('lat', 'lon'),
                           coords={'lat': ds['lat'], 'lon': ds['lon']})
    return da_mask
```

```python
mat = loadmat("Heat_budget.mat")

# Budget terms and regions present in the file
terms   = ['rate', 'forc', 'xadv', 'yadv', 'vadv', 'hmix', 'vmix', 'entr', 'temp']
regions = ['coast', 'off']

# --- time axis: 432 monthly steps ---
# !!! set the real start date of your run (guess below: Jan 1980) ---
time = pd.date_range("1980-01-01", periods=432, freq="MS")

# Stack each term into a (time, region) array
data_vars = {}
for t in terms:
    cols = [np.asarray(mat[f"BUDGET_{r}_{t}"]).ravel() for r in regions]
    data_vars[t] = (("time", "region"), np.column_stack(cols).astype("float64"))

ds = xr.Dataset(
    data_vars,
    coords={"time": time, "region": regions},
)

# Optional: human-readable descriptions
long_names = {
    'rate': 'MLT tendency (dT/dt)', 'forc': 'Surface forcing',
    'xadv': 'Zonal advection',      'yadv': 'Meridional advection',
    'vadv': 'Vertical advection',   'hmix': 'Horizontal mixing',
    'vmix': 'Vertical mixing',      'entr': 'Entrainment',
    'temp': 'Mixed-layer temperature',
}
for k, v in long_names.items():
    ds[k].attrs['long_name'] = v

ds
```

```python
from scipy.stats import pearsonr

# ---- seasonal cycle of the coastal heat budget ----
clim = ds.groupby("time.month").mean().sel(region="coast")
months = clim.month.values

# Upwelling signature in the budget (cooling from below).
# Default = vertical advection + entrainment. Add 'vmix' if your turbulent
# flux at the ML base represents entrained cold water (see note below).
upw_terms = ["vadv", "entr"]            # try ["vadv","entr","vmix"]
upw_cool  = sum(clim[t] for t in upw_terms).values     # negative (cooling)
forc      = clim["forc"].values                        # surface forcing (warming)
net       = clim["rate"].values                        # net SST tendency (ΔT)

# Plot the COOLING MAGNITUDE so it's positive and directly comparable in
# phase to the (positive) wind-driven upwelling indices.
upw_mag = -upw_cool

# ---- OPTIONAL overlays: provide 12-month climatologies you already have ----
# from diagnostics 1 & 2 (set to None to skip)
WMLD_clim = monthly_mean(wmld).mean(dim=('lat','lon')).WMLD     # e.g. wmld_seasonal.values   (length 12, Jan..Dec)
CUI_clim  = monthly_mean(CUI).mean('lat').CUI_cc     # e.g. cui_seasonal.values

fig, ax = plt.subplots(figsize=(8.5, 5))

# Budget terms (left axis, °C/d)
ax.plot(months, upw_mag, color="royalblue", lw=2.5, marker="o", ms=4,
        label=r"Upwelling cooling  $-(\mathrm{vadv}+\mathrm{entr})$")
ax.plot(months, forc, color="mediumpurple", lw=2.5, marker="s", ms=4,
        label=r"Surface forcing (forc)")
ax.plot(months, net, color="black", lw=2.5, marker="d", ms=4,
        label=r"Net tendency $\Delta T$ (rate)")
ax.axhline(0, color="grey", lw=0.6)
ax.set_ylabel(r"ML heat budget ($^{\circ}$C d$^{-1}$)", fontsize=12)
ax.set_xlim(1, 12); ax.set_xticks(range(1, 13))
ax.set_xticklabels(list("JFMAMJJASOND"))
ax.tick_params(direction="in")

# Reference overlays (right axis, normalized 0–1 just for phase comparison)
handles, labels = ax.get_legend_handles_labels()
if WMLD_clim is not None or CUI_clim is not None:
    axr = ax.twinx()

    if WMLD_clim is not None:
        l, = axr.plot(months, norm(WMLD_clim), color="dimgrey", lw=2,
                      ls="--", marker="^", ms=4, label=r"$W_{MLD}$ (norm.)")
        handles.append(l)
    if CUI_clim is not None:
        l, = axr.plot(months, norm(CUI_clim), color="red", lw=2,
                      ls="--", marker="v", ms=4, label="CUI (norm.)")
        handles.append(l)
    axr.set_ylabel("normalized index", fontsize=11)
    # axr.set_ylim(-0.05, 1.05)

ax.legend(handles=handles, fontsize=9, frameon=False, loc="upper center", ncol=2)
ax.set_title("Coastal: upwelling cooling tracks the winds; "
             "net SST tendency does not", fontsize=11, fontweight="bold")
fig.tight_layout()
# fig.savefig("Figures/budget_decomposition_coast.pdf", bbox_inches="tight", dpi=300)
plt.show()

# ---- quantify the argument ----
def corr(a, b):
    a, b = np.asarray(a, float), np.asarray(b, float)
    m = np.isfinite(a) & np.isfinite(b)
    return pearsonr(a[m], b[m])

if WMLD_clim is not None:
    r1, p1 = corr(upw_mag, WMLD_clim)   # expect HIGH positive
    r2, p2 = corr(net,     WMLD_clim)   # expect LOW / not significant
    print(f"corr( upwelling cooling , WMLD ) = {r1:+.2f}  (p={p1:.2g})")
    print(f"corr( net tendency ΔT   , WMLD ) = {r2:+.2f}  (p={p2:.2g})")
if (WMLD_clim is not None) and (CUI_clim is not None):
    r3, p3 = corr(CUI_clim, WMLD_clim)  # expect LOW / negative
    r4, p4 = corr(CUI_clim, net)        # CUI closer to net than to forcing
    print(f"corr( CUI , WMLD )              = {r3:+.2f}  (p={p3:.2g})")
    print(f"corr( CUI , net tendency ΔT )   = {r4:+.2f}  (p={p4:.2g})")
```

```python
def sym_lim(mean, factor=1.4):
    hw = abs(mean) * factor
    return (mean - hw, mean + hw)
    
def rr(a, b):
    a, b = np.asarray(a, float), np.asarray(b, float)
    m = np.isfinite(a) & np.isfinite(b)
    return pearsonr(a[m], b[m])[0]

# --- figure with 3 y-axes ---
fig, ax1 = plt.subplots(figsize=(8.5, 5.2))
ax2 = ax1.twinx()
ax3 = ax1.twinx()

# offset the third axis to the right
ax3.spines['right'].set_position(('axes', 1.15))

# colors matching each variable
c1, c2, c3 = 'royalblue', 'seagreen', 'dimgrey'

# --- plot each on its own axis ---
l1, = ax1.plot(months, upw_mag, color=c1, lw=3, marker='o', ms=6,
               label='Upwelling cooling  $-(\\mathrm{vadv}+\\mathrm{entr})$')
l2, = ax2.plot(months, 1/MLD, color=c2, lw=2.5, ls='--', marker='s', ms=6,
               label='Stratification proxy  $1/\\mathrm{MLD}$')
l3, = ax3.plot(months, WMLD_clim.data, color=c3, lw=2.5, ls=':', marker='^', ms=7,
               label='Upwelling velocity  $W_{MLD}$')

# --- y-axis labels colored to match ---
ax1.set_ylabel('Upwelling cooling [°C/day]', color=c1)
ax2.set_ylabel('1/MLD [m$^{-1}$]', color=c2)
ax3.set_ylabel('$W_{MLD}$ [m/s]', color=c3)

# color the tick labels
ax1.tick_params(axis='y', colors=c1, direction='in', size=6, width=1.3)
ax2.tick_params(axis='y', colors=c2, direction='in', size=6, width=1.3)
ax3.tick_params(axis='y', colors=c3, direction='in', size=6, width=1.3)

# set ylim
ax1.set_ylim(sym_lim(np.mean(upw_mag)))
ax2.set_ylim(sym_lim(np.mean(1/MLD)))
ax3.set_ylim(sym_lim(np.mean(WMLD_clim.data)))

# color the spines
ax1.spines['left'].set_color(c1)
ax2.spines['right'].set_color(c2)
ax3.spines['right'].set_color(c3)

# --- x-axis ---
ax1.set_xlim(1, 12)
ax1.set_xticks(range(1, 13))
ax1.set_xticklabels(list("JFMAMJJASOND"))
ax1.set_xlabel('month')
ax1.tick_params(axis='x', direction='in', size=6, width=1.3)
ax1.grid(True, color='lightgrey', lw=0.4)
ax1.set_axisbelow(True)

# --- combined legend ---
lines = [l1, l2, l3]
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='lower right', fontsize=11, frameon=False)

# --- correlation box ---
ax1.text(0.025, 0.04,
         f"$r$(cooling, $1/$MLD) = {rr(upw_mag, strat):+.2f}\n"
         f"$r$(cooling, $W_{{MLD}}$) = {rr(upw_mag, Wv):+.2f}",
         transform=ax1.transAxes, va='bottom', ha='left', fontsize=12,
         bbox=dict(boxstyle='round', fc='white', ec='grey', alpha=0.9))

ax1.set_title('Cooling in the mixed layer is controled by 1/MLD, not upwelling velocity',
              fontsize=12, fontweight='bold')

fig.tight_layout()
# fig.savefig('Figures/paper/CUI_stratification_mechanism.pdf', bbox_inches='tight', dpi=300)
plt.show()
```

```python
plt.rcParams.update({
    'font.size':        12,
    'axes.titlesize':   13,
    'axes.labelsize':   12,
    'xtick.labelsize':  11,
    'ytick.labelsize':  11,
    'legend.fontsize':  10,
})

term_colors = {
    'rate':  'black',   'forc':  'mediumpurple', 'xadv': 'steelblue',
    'yadv':  'goldenrod','vadv': 'forestgreen',  'vmix': 'tab:red',
    'entr':  'chocolate',
}
term_labels = {
    'rate':'$\\Delta T$','forc':'forc','xadv':'xadv','yadv':'yadv',
    'vadv':'vadv','vmix':'vmix','entr':'entr',
}

def monthly_clim(ds):
    g = ds.groupby('time.month')
    return g.mean(), g.std()

def plot_terms(ax, mean_ds, std_ds, colors, labels, title):
    months_ = mean_ds['month'].values
    for var, color in colors.items():
        if var not in mean_ds.data_vars:
            continue
        m = mean_ds[var].values
        s = std_ds[var].values
        ax.plot(months_, m, color=color, lw=2, label=labels[var])
        ax.fill_between(months_, m - s, m + s, color=color, alpha=0.2, linewidth=0)
    ax.axhline(0, color='gray', ls='--', lw=1)
    ax.set_title(title)
    ax.set_xticks(range(1, 13))
    ax.set_xlim([1, 12])
    ax.set_xticklabels(list("JFMAMJJASOND"))
    ax.grid(True, ls='--', alpha=0.5)

def sym_lim(mean, factor=1.4):
    hw = abs(mean) * factor
    return (mean - hw, mean + hw)

def rr(a, b):
    a, b = np.asarray(a, float), np.asarray(b, float)
    m = np.isfinite(a) & np.isfinite(b)
    return pearsonr(a[m], b[m])[0]

# --- Heat budget climatology (Coastal - Offshore) ---
ds_clean = ds.drop_vars(['hmix', 'temp'])
diff_m, diff_s = monthly_clim(ds_clean.sel(region='coast') - ds_clean.sel(region='off'))
for d in (diff_m, diff_s):
    for v in d.data_vars:
        d[v] = d[v] * 86400

# --- Map data ---
proj = ccrs.PlateCarree()
step = 7
sst_mean = Temp_mld.Temp.mean('time').sel(lat=slice(-20, 0), lon=slice(-90, -70))
uv_mean = (
    UV_mld.mean('time')
    .sel(lon=slice(-90, -70), lat=slice(-20, 0))
    .coarsen(lon=step, lat=step, boundary="trim")
    .mean()
)

# =================== Combined figure ===================
fig = plt.figure(figsize=(5*2, 4*1.5))
gs = fig.add_gridspec(
    2, 2,
    width_ratios=[1, 1.5],
    height_ratios=[1.5, 1.5],     # B:C = 1:3
    wspace=0.32, hspace=0.30,
)

# ---------- Panel A: heat budget (spans both rows) ----------
ax_A = fig.add_subplot(gs[:, 0])
plot_terms(ax_A, diff_m, diff_s, term_colors, term_labels, "(a) Coastal − Offshore")
ax_A.set_ylabel('[°C/day]')
handles, labels = ax_A.get_legend_handles_labels()
ax_A.legend(handles, labels, ncol=4,
            loc='upper center', bbox_to_anchor=(0.5, -0.08), frameon=False)

# ---------- Panel B: 3-axis cooling / MLD (top right) ----------
ax1 = fig.add_subplot(gs[0, 1])
ax2 = ax1.twinx()
ax3 = ax1.twinx()
ax3.spines['right'].set_position(('axes', 1.18))

c1, c2, c3 = 'royalblue', 'seagreen', 'dimgrey'

l1, = ax1.plot(months, upw_mag, color=c1, lw=3, marker='o', ms=5,
               label='Upwelling cooling  $-(\\mathrm{vadv}+\\mathrm{entr})$')
l2, = ax2.plot(months, 1/MLD, color=c2, lw=2.5, ls='--', marker='s', ms=5,
               label='Stratification proxy  $1/\\mathrm{MLD}$')
l3, = ax3.plot(months, WMLD_clim.data, color=c3, lw=2.5, ls=':', marker='^', ms=6,
               label='$W_{MLD}$')

ax1.set_ylabel('Cooling [°C/day]', color=c1, fontsize=10)
ax2.set_ylabel('1/MLD [m$^{-1}$]', color=c2, fontsize=10)
ax3.set_ylabel('$W_{MLD}$ [m/s]', color=c3, fontsize=10)

ax1.tick_params(axis='y', colors=c1, direction='in', size=5, width=1.2, labelsize=9)
ax2.tick_params(axis='y', colors=c2, direction='in', size=5, width=1.2, labelsize=9)
ax3.tick_params(axis='y', colors=c3, direction='in', size=5, width=1.2, labelsize=9)

ax1.set_ylim(sym_lim(np.mean(upw_mag)))
ax2.set_ylim(sym_lim(np.mean(1/MLD)))
ax3.set_ylim(sym_lim(np.mean(WMLD_clim.data)))

ax1.spines['left'].set_color(c1)
ax2.spines['right'].set_color(c2)
ax3.spines['right'].set_color(c3)

ax1.set_xlim(1, 12)
ax1.set_xticks(range(1, 13))
ax1.set_xticklabels(list("JFMAMJJASOND"))
ax1.tick_params(axis='x', direction='in', size=5, width=1.2, labelsize=9)
ax1.grid(True, color='lightgrey', lw=0.4)
ax1.set_axisbelow(True)

lines = [l1, l2, l3]
ax1.legend(lines, [l.get_label() for l in lines],
           loc='upper right', fontsize=8, frameon=False)

ax1.text(0.025, 0.06,
         f"$r$(cooling, $1/$MLD) = {rr(upw_mag, strat):+.2f}\n"
         f"$r$(cooling, $W_{{MLD}}$) = {rr(upw_mag, Wv):+.2f}",
         transform=ax1.transAxes, va='bottom', ha='left', fontsize=9,
         bbox=dict(boxstyle='round', fc='white', ec='grey', alpha=0.9))

# ax1.set_title('(b) Cooling controlled by 1/MLD, not upwelling velocity',
#               fontsize=11)

# ---------- Panel C: map (bottom right, bigger) ----------
ax_C = fig.add_subplot(gs[1, 1], projection=proj)
sst_mean.plot.contourf(
    x="lon", y="lat",
    vmin=12, vmax=28, levels=17,
    cmap=cm.cm.balance,
    transform=proj, ax=ax_C,
    cbar_kwargs=dict(shrink=0.7, label="[°C]"),
)
Q = ax_C.quiver(
    uv_mean.lon, uv_mean.lat,
    uv_mean.u.transpose("lat", "lon").values,
    uv_mean.v.transpose("lat", "lon").values,
    transform=proj, scale=5, width=0.0035, color="k",
)
ax_C.quiverkey(Q, 0.85, 0.35, 0.2, "0.2 m/s",
               labelpos='S', coordinates='axes',
               fontproperties=dict(size=10))

ax_C.plot(coast_lons + 0.05, lat_all, color='black', lw=1, label='Coast')
ax_C.plot(offshore_lons - 0.05, lat_all, color='black', lw=1, label='Offshore')

for lat0 in coastal_b["target_lat"].values:
    if not coastal_b["transect_valid"].sel(target_lat=lat0).item():
        continue
    x0 = coastal_b["transect_x0"].sel(target_lat=lat0).item()
    x1 = coastal_b["transect_x1"].sel(target_lat=lat0).item()
    ax_C.plot([x0, x1], [lat0 + 0.05, lat0 + 0.05], color='black', lw=1, zorder=5)

ax_C.add_feature(cfeature.LAND, facecolor="0.7")
ax_C.set_xticks([-90, -85, -80, -75, -70], crs=proj)
ax_C.set_yticks([-20, -15, -10, -5, 0], crs=proj)
ax_C.xaxis.set_major_formatter(LongitudeFormatter())
ax_C.yaxis.set_major_formatter(LatitudeFormatter())
ax_C.set_xlabel("")
ax_C.set_ylabel("")
# ax_C.set_title("(c) Temperature and Currents within MLD")

# fig.savefig('figures/heat_budget_combined.png', bbox_inches='tight', dpi=300)
plt.show()
```
