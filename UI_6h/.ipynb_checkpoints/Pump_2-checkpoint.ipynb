```python
import xarray as xr
```

```python
import glob

files = sorted(glob.glob('Pump/data/Pump*.nc'))
print(len(files), files[:5], files[-5:])

# Peek at first two
for f in files[:3]:
    with xr.open_dataset(f) as ds:
        print(f, ds['time'].load().values[:3], ds['time'].load().values[-3:])
```

```python
ds = xr.open_mfdataset(
    files,
    combine='nested',
    concat_dim='time',  # name of the time coordinate/dimension to concat
    parallel=True,
    chunks=
        'auto'# Half the spatial dimension
).sortby('time')
ds = ds.compute()
ds
```

```python
ds = ds.dropna('lat')
ds.w_ek.plot.contourf(robust=True,x='time')
```

```python
import matplotlib.pyplot as plt

ds.w_ek.mean(dim='lat').plot(figsize=(10,5))
plt.grid()
```

```python
ds.w_ek.groupby('time.month').mean().plot.contourf(x='month', robust=True)
plt.title('Seasonality')
```

```python
ds.w_ek.resample(time='1M').mean().rolling(lat=13, center=True).mean().plot.contourf(x='time',robust=True)
```

```python
ds.w_ek.resample(time='1M').mean().mean(dim='lat').rolling(time=13, center=True).mean().rename('$m^2 \cdot s^{-1}$').plot()
plt.grid()
```
