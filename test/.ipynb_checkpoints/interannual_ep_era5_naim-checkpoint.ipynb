```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
```

```python
winds = xr.open_dataset('era5_jja_1982-2023.nc').squeeze(drop=True)
dlsm = xr.open_dataset('era5_land_sea_mask.nc', decode_times=False).squeeze(drop=True)
winds
```

```python
dlsm = dlsm.rename({'longitude': 'lon', 'latitude': 'lat'})
winds = winds.rename({'longitude': 'lon', 'latitude': 'lat'})
```

```python
lon = winds.lon #check names
lat = winds.lat
uwind = winds.u10 #.isel(time=range(0,42))
vwind = winds.v10 #.isel(time=range(0,42))
lsm = dlsm.lsm
uwind
```

```python
lsm1 = xr.where(lsm > 0.5, 1, 0)
lsm1.plot()
```

```python
u_wind_masked = uwind.where(lsm1 == 0)
v_wind_masked = vwind.where(lsm1 == 0)
u10 = u_wind_masked.mean('time')
v10 = v_wind_masked.mean('time')
```

```python
plt.quiver(lon,lat,uwind[1],vwind[1])
```

```python
#plt.quiver(lon,lat,u10[1],v10[1])
```

```python
ws = np.sqrt(u10**2 + v10**2)
omega = 7.29e-5
f = 2 * omega * np.sin(np.radians(lat))
pa = 1.22
cd = 1.3e-3
taux = pa * cd * ws * u10
tauy = pa * cd * ws * v10
```

```python
A = 6371.0e3
```

```python
delta_y = np.deg2rad(winds.lat.shift(lat=-1) - winds.lat) * A
#delta_y
```

```python
delta_x = (
    np.deg2rad(winds.lon.shift(lon=-1) - winds.lon) * A
    * np.cos(np.deg2rad(winds.lat))
)
#delta_x
```

```python
dtauy_by_dx = (tauy.shift(lon=-1) - tauy) / delta_x
dtauy_by_dx = (dtauy_by_dx.shift(lat=-1) + dtauy_by_dx) / 2.0
#dtauy_by_dx
```

```python
dtaux_by_dy = (taux.shift(lat=-1) - taux) / delta_y
dtaux_by_dy = (dtaux_by_dy.shift(lon=-1) + dtaux_by_dy) / 2.0
#dtaux_by_dy
```

```python
lon1 = (lon.shift(lon=-1) + lon) / 2
lat1 = (lat.shift(lat=-1) + lat) / 2
```

```python
dtaux_by_dy = dtaux_by_dy.assign_coords(lon = lon1, lat = lat1).isel(lon=slice(None, -1),lat=slice(None, -1))
dtauy_by_dx = dtauy_by_dx.assign_coords(lon = lon1, lat = lat1).isel(lon=slice(None, -1),lat=slice(None, -1))
```

```python
curlz_tau = (dtauy_by_dx - dtaux_by_dy)
curlz_tau
```

```python
f = 2 * 7.2921e-5 * np.sin(np.deg2rad(curlz_tau.lat))
rho = 1000
f.shape
```

```python
import cmocean
import cartopy
from matplotlib import pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cft
```

```python
w_ek = (1/f/rho) * curlz_tau
w_ek
```

```python
w_ek = w_ek.transpose('valid_time','lon','lat')
```

```python
ndelta_y = -(np.deg2rad(curlz_tau.lat.shift(lat=-1) - curlz_tau.lat) * A)
ndelta_y
```

```python
ndelta_x = (
    np.deg2rad(curlz_tau.lon.shift(lon=-1) - curlz_tau.lon) * A
    * np.cos(np.deg2rad(curlz_tau.lat))
)
ndelta_x
```

```python
da = ndelta_x * ndelta_y
da
```

```python
total_ep = w_ek * da
total_ep
```
